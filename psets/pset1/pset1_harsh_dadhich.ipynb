Looking at IBM's efficiency through two lenses:
- **Total Asset Turnover** — how well they use assets to generate revenue
- **Inventory Turnover** — how quickly they move through inventory

Goal: see if IBM got better or worse at these between 2022 and 2023.

## imports

```python
import yfinance as yf
import pandas as pd

pd.set_option("display.float_format", lambda x: "%.2f" % x)
```

## fetching IBM data

```python
# grab ibm ticker and pull financial statements
ibm = yf.Ticker("IBM")

balance_sheet = ibm.balance_sheet
income_statement = ibm.income_stmt

print("balance sheet:")
display(balance_sheet)
```

```python
print("income statement:")
display(income_statement)
```

## extracting the numbers we need

from the balance sheet we need total assets and inventory (2021-2023 for avg calculations), and from income stmt we need revenue and COGS for 2022/2023.

```python
# check what years we have available
print("years in balance sheet:")
for col in balance_sheet.columns:
    print(f"  {col.year}")

print("\nyears in income stmt:")
for col in income_statement.columns:
    print(f"  {col.year}")
```

```python
# helper to grab column by year (yfinance cols are datetime objs)
def get_col_by_year(df, year):
    for col in df.columns:
        if col.year == year:
            return col
    return None


# pull total assets 2021-2023
total_assets_2021 = balance_sheet.loc[
    "Total Assets", get_col_by_year(balance_sheet, 2021)
]
total_assets_2022 = balance_sheet.loc[
    "Total Assets", get_col_by_year(balance_sheet, 2022)
]
total_assets_2023 = balance_sheet.loc[
    "Total Assets", get_col_by_year(balance_sheet, 2023)
]

# pull inventory 2021-2023
inventory_2021 = balance_sheet.loc["Inventory", get_col_by_year(balance_sheet, 2021)]
inventory_2022 = balance_sheet.loc["Inventory", get_col_by_year(balance_sheet, 2022)]
inventory_2023 = balance_sheet.loc["Inventory", get_col_by_year(balance_sheet, 2023)]

# pull revenue 2022-2023
revenue_2022 = income_statement.loc[
    "Total Revenue", get_col_by_year(income_statement, 2022)
]
revenue_2023 = income_statement.loc[
    "Total Revenue", get_col_by_year(income_statement, 2023)
]

# pull cogs 2022-2023
cogs_2022 = income_statement.loc[
    "Cost Of Revenue", get_col_by_year(income_statement, 2022)
]
cogs_2023 = income_statement.loc[
    "Cost Of Revenue", get_col_by_year(income_statement, 2023)
]

# print extracted vals (in millions)
print("extracted data (millions USD):\n")
print(f"total assets:")
print(f"  2021: ${total_assets_2021/1e6:,.2f}M")
print(f"  2022: ${total_assets_2022/1e6:,.2f}M")
print(f"  2023: ${total_assets_2023/1e6:,.2f}M")

print(f"\ninventory:")
print(f"  2021: ${inventory_2021/1e6:,.2f}M")
print(f"  2022: ${inventory_2022/1e6:,.2f}M")
print(f"  2023: ${inventory_2023/1e6:,.2f}M")

print(f"\nrevenue:")
print(f"  2022: ${revenue_2022/1e6:,.2f}M")
print(f"  2023: ${revenue_2023/1e6:,.2f}M")

print(f"\ncogs:")
print(f"  2022: ${cogs_2022/1e6:,.2f}M")
print(f"  2023: ${cogs_2023/1e6:,.2f}M")
```

## calculating the ratios

formulas:
- **total asset turnover** = revenue / avg total assets
- **inventory turnover** = cogs / avg inventory

```python
# avg total assets (beginning + ending / 2)
avg_total_assets_2022 = (total_assets_2021 + total_assets_2022) / 2
avg_total_assets_2023 = (total_assets_2022 + total_assets_2023) / 2

# avg inventory
avg_inventory_2022 = (inventory_2021 + inventory_2022) / 2
avg_inventory_2023 = (inventory_2022 + inventory_2023) / 2

print("averages (millions USD):\n")
print(f"avg total assets:")
print(f"  2022: ${avg_total_assets_2022/1e6:,.2f}M")
print(f"  2023: ${avg_total_assets_2023/1e6:,.2f}M")

print(f"\navg inventory:")
print(f"  2022: ${avg_inventory_2022/1e6:,.2f}M")
print(f"  2023: ${avg_inventory_2023/1e6:,.2f}M")
```

```python
# total asset turnover = revenue / avg assets
total_asset_turnover_2022 = revenue_2022 / avg_total_assets_2022
total_asset_turnover_2023 = revenue_2023 / avg_total_assets_2023

# inventory turnover = cogs / avg inventory
inventory_turnover_2022 = cogs_2022 / avg_inventory_2022
inventory_turnover_2023 = cogs_2023 / avg_inventory_2023

print("results:\n")
print(f"total asset turnover:")
print(f"  2022: {total_asset_turnover_2022:.4f}")
print(f"  2023: {total_asset_turnover_2023:.4f}")

print(f"\ninventory turnover:")
print(f"  2022: {inventory_turnover_2022:.4f}")
print(f"  2023: {inventory_turnover_2023:.4f}")
```

## comparing 2022 vs 2023

let's see how these changed year over year.

```python
# yoy changes
tat_change = total_asset_turnover_2023 - total_asset_turnover_2022
tat_pct_change = (
    (total_asset_turnover_2023 - total_asset_turnover_2022) / total_asset_turnover_2022
) * 100

inv_change = inventory_turnover_2023 - inventory_turnover_2022
inv_pct_change = (
    (inventory_turnover_2023 - inventory_turnover_2022) / inventory_turnover_2022
) * 100

# summary table
comparison_data = {
    "Ratio": ["Total Asset Turnover", "Inventory Turnover"],
    "2022": [total_asset_turnover_2022, inventory_turnover_2022],
    "2023": [total_asset_turnover_2023, inventory_turnover_2023],
    "Change": [tat_change, inv_change],
    "% Change": [tat_pct_change, inv_pct_change],
    "Trend": [
        "Improved" if tat_change > 0 else "Worsened",
        "Improved" if inv_change > 0 else "Worsened",
    ],
}

comparison_df = pd.DataFrame(comparison_data).set_index("Ratio")
display(comparison_df)
```

```python
# breakdown
print("total asset turnover:")
print(f"  2022: {total_asset_turnover_2022:.4f}")
print(f"  2023: {total_asset_turnover_2023:.4f}")
print(f"  change: {tat_pct_change:+.2f}%")
print(f"  verdict: {'improved' if tat_change > 0 else 'worsened'}")

print(f"\ninventory turnover:")
print(f"  2022: {inventory_turnover_2022:.4f}")
print(f"  2023: {inventory_turnover_2023:.4f}")
print(f"  change: {inv_pct_change:+.2f}%")
print(f"  verdict: {'improved' if inv_change > 0 else 'worsened'}")
```

## Observations

**Total asset turnover** Went from 0.4670 to 0.4713 (+0.94%). Small uptick - IBM squeezed a bit more revenue out of each dollar in assets. not a dramatic shift, but moving in the right direction.

**Inventory turnover** jumped from 17.40 to 20.32 (+16.79%). This is the standout. IBM's moving through inventory ~17% faster than last year. they cut avg inventory from $1.6B to $1.36B while keeping cogs roughly flat - that's tighter ops.

**Bottom line:** Both metrics improved yoy. the inventory turnover gain is particularly strong and suggests IBM's gotten leaner on the supply chain side. asset turnover is relatively flat but stable - for a $135B asset base, that's not bad.

**What does this mean for IBM?** They're getting more efficient, especially on the inventory side. the asset turnover staying flat isn't ideal, but the inventory gains suggest they're tightening up operations where they can. overall, these ratios paint a picture of a company that's improving its efficiency year over year - nothing explosive, but solid progress.
